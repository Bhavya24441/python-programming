Question:

Create a class ATM to verify PIN and allow access.

```python
#Soultion:

class ATM:
    def __init__(self,pin):
        self.pin = pin
    def verify(self,input_pin):
        if input_pin == self.pin:
            print("Access Granted!")
        else:
            print("Wrong pin")
a = ATM(1234)
a.verify(12455)
```

Question:

Create a class Calculator to perform addition and multiplication.

```python
#solution:
class Calculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def add(self):
        return self.a + self.b
    def multiply(self):
        return self.a *self.b
c = Calculator(23,45)
print(c.add())
print(c.multiply())
```

Question:

Create a class Laptop to turn on/off and show status.

```python
#Solution:
class Laptop:
    def __init(self):
        self.on = False
    def power_on(self):
        self.on = True
    def power_off(self):
        self.on = False
    def status(self):
        print("On " if self.on else "off")
l = Laptop()
l.power_on()
l.status()
```

Question:

Create a class Course where students can enroll and total count is tracked.

```python
#Solution:
class Course:
    def __init__(self,name):
        self.name = name
        self.students = []
    def enroll(self,student):
        self.students.append(student)
    def total(self):
        print(len(self.students))

c = Course("Python")

c.enroll("B")
c.total()
```
