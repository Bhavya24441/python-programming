Question:

Create a class ParkingLot to manage available slots.

```python
#Solution:
class ParkingLot:
    def __init__(self, capacity):
        self.capacity = capacity
        self.occupied = 0
    def park(self):
        if self.occupied < self.capacity:
            self.occupied +=1
            print("Parked")
        else:
            print("Full")
    def leave(self):
        if self.occupied > 0:
            self.occupied -=1
            print("Left")
p = ParkingLot(2)
p.park()
p.park()
p.park()
p.leave()
```

Question:

Create a class Quiz to track correct answers and score.

```python
#Solution:
class Quiz:
    def __init__(self):
        self.score = 0
    def correct(self):
        self.score +=1
    def show_score(self):
        print(self.score)
q = Quiz()
q.correct()
q.correct()
q.show_score()
```

Question:

Create a class File to check if file size exceeds limit.

```python
#Solution:
class File:
    def __init__(self, size):

        self.size = size 
    def is_large(self):
        return self.size >100
f = File(120)
print(f.is_large())
```

Question:

Create a class Light to toggle on/off state.

```python
#Solution:
class Light:
    def __init__(self):
        self.state = False
    def toggle(self):
        self.state = not self.state
    def show(self):
        print("On" if self.state else " off")
l = Light()
l.toggle()
l.show()
```
