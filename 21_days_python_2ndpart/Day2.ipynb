Question:

Create a class Book to manage issuing and returning a book.

```python
#Solution:
class Book:
    def __init__(self,title):
        self.title = title 
        self.available = True
    
    def issue(self):
        if self.available:
            self.available = False
            print("Issued")
        else:
            print("Not available")
    def return_book(self):
        self.available = True
        print("Returned")
b = Book("Python")
b.issue()
b.return_book()
```

Question:

Create a class BankAccount to perform deposit and withdraw operations safely.

```python
#Solution:
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self,amount):
        self.balance+=amount
        print(self.balance)
    def withdraw(self, amount):
        if amount <=self.balance:
            self.balance -= amount
            print(self.balance)
        else:
            print("Insufficient balance")

acc = BankAccount(1000)
acc.deposit(500)
acc.withdraw(300)
```

Question:

Create a class Student to calculate grade based on marks.

```python
#Solution:
class Student:
    def __init__(self,marks):
        self.marks = marks
    def grade(self):
        if self.marks >=90:
            return "A"
        elif self.marks >= 75:
            return "B"
        else:
            return "C"
s = Student(80)
print(s.grade())
```

Question:

Create a class Cart to add items and calculate total price.

```python
#Solution:
class Cart:
    def __init__(self):
        self.total = 0
    def add_item(self,price):
        self.total += price
    def show_total(self):
        print(self.total)
c = Cart()
c.add_item(100)
c.add_item(200)
c.show_total()
```
