Question:

Create a class Product with price and apply discount method.

```python
#Solution:
class Product:
    def __init__(self,price):
        self.price = price
    def apply_discount(self, percent):
        self.price = self.price - (self.price * percent/100)
    def show_price(self):
        print(self.price)
p = Product(1000)
p.apply_discount(10)
p.show_price()
```

Question:

Create a class Circle to calculate area and perimeter.

```python
#Solution:
class Circle:
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius 
    def perimeter(self):
        return 2* 3.14*self.radius
c = Circle(5)
print(c.area())
print(c.perimeter())
```

Question:

Create a class Password to check if password length is strong (>=8).

```python
#Solution:
class Password:
    def __init__(self, pwd):
        self.pwd = pwd
    def is_strong(self):
        return len(self.pwd) >=8
p = Password("alkdfja23334")
print(p.is_strong())
```

Question:

Create a class Movie to store rating and classify it.

```python
#Solution:
class Movie:
    def __init__(self, rating):
        self.rating = rating
    def category(self):
        if self.rating>=8:
            return "Hit"
        elif self.rating >=5:
            return "Average"
        else:
            return "Flop"
m = Movie(9)
print(m.category())

        
```
