```python
#Answer3: 
def missing_number(nums):
    n = len(nums)
    total = n * (n + 1) / 2
    return total - sum(nums)
nums = [3, 0, 1]
print(missing_number(nums))
```

```python
#Answer4: 
def add_unique(x, seen=set()):
    if seen in None:
        seen =set()
    if x not in seen:
        seen.add(x)
    return list(seen)
print(add_unique(1))
print(add_unique(2))
```
