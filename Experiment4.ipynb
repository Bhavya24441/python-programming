Q1) Find a factorial of given number. 

```python
#Solution:
n = int(input("enter a number: "))
fact = 1 
for i in range(1, n+1):
    fact = fact * 1 
print("Factorial= ", fact)
```

Q2) Find whether the given number is Armstrong number.  


```python
#Solution:
num = int(input("Enter a number: "))
temp = num
sum = 0
while temp> 0:
    digit = temp % 10 
    sum = sum + digit **3 
    temp = temp // 10
if sum == num:
    print("Armstrong number")
else:
    print("Not a armstrong number")
    
```

Q4) Write a program to find if given number is prime number or not.  

```python
#Solution:
num = int(input("Enter a number: "))
if num > 1:
    for i in range(2,num):
        if num % i == 0:
            print("Not a prime number")
            break
        else:
            print("It is a prime number")
    else:
        print("not a prime number")
```

Q5) Check whether given number is palindrome or not. 

```python
#Solution:
num = int(input("Enter a number:"))
temp = num
rev = 0
while temp > 0:
    digit = temp % 10 
    rev = rev * 10 + digit 
    temp = temp // 10
if rev == num:
    print("It is a palindrome number")
else:
    print("Not a palindrome number") 
```
